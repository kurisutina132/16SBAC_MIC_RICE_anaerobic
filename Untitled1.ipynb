Limpieza: pasar "NA", "NULL", "" a nulos reales


```python
from pyspark.sql import functions as F

missing = ["NA", "NULL", "null", ""]

df0 = df
for c in df0.columns:
    df0 = df0.withColumn(
        c,
        F.when(F.trim(F.col(c)).isin(missing), F.lit(None)).otherwise(F.col(c))
    )
```

Convertir fecha_hora a timestamp y las mediciones a double

```python
df_clean = (
    df0
    .withColumn("fecha_hora", F.to_timestamp("fecha_hora", "yyyy-MM-dd HH:mm:ss"))
    .withColumn("temperatura", F.col("temperatura").cast("double"))
    .withColumn("humedad", F.col("humedad").cast("double"))
    .withColumn("ph", F.col("ph").cast("double"))
    .withColumn("oxigeno_dis", F.col("oxigeno_dis").cast("double"))
    .withColumn("conductividad", F.col("conductividad").cast("double"))
    .withColumn("turbidez", F.col("turbidez").cast("double"))
)
```

```python
from pyspark.sql import functions as F

# patrón: 2024-03-16 13:45:00
pat_ts = r"^\d{4}-\d{2}-\d{2} \d{2}:\d{2}:\d{2}$"

df_qc = (df0
    .withColumn("fecha_hora_raw", F.col("fecha_hora"))  # guardamos original
    .withColumn(
        "fecha_hora",
        F.when(F.col("fecha_hora_raw").rlike(pat_ts),
               F.to_timestamp("fecha_hora_raw", "yyyy-MM-dd HH:mm:ss")
        ).otherwise(F.lit(None))
    )
    .withColumn("temperatura", F.col("temperatura").cast("double"))
    .withColumn("humedad", F.col("humedad").cast("double"))
    .withColumn("ph", F.col("ph").cast("double"))
    .withColumn("oxigeno_dis", F.col("oxigeno_dis").cast("double"))
    .withColumn("conductividad", F.col("conductividad").cast("double"))
    .withColumn("turbidez", F.col("turbidez").cast("double"))
)
```

Reglas de rangos (agua) + separar válidos/ inválidos

Rangos sugeridos:

temperatura (°C): 0 a 40 (si hay agua fría, podrías bajar a -2)
pH: 0 a 14 (más estricto típico 4 a 10, pero depende)
oxígeno disuelto (mg/L): 0 a 20
conductividad (µS/cm): 0 a 10000 (depende muchísimo, pero negativos seguro no)
turbidez (NTU): 0 a 3000 (negativos no; máximos dependen del río/obra)

```python
df_ok = df_ok_fecha
```

```python
rules = {
    "temperatura": (0.0, 40.0),
    "ph": (0.0, 14.0),
    "oxigeno_dis": (0.0, 20.0),
    "conductividad": (0.0, 10000.0),
    "turbidez": (0.0, 3000.0),
    # humedad: si es “humedad ambiente” cerca del agua, 0..100.
    "humedad": (0.0, 100.0),
}

def out_of_range(colname, lo, hi):
    # null no cuenta como raro aquí (solo “faltante”)
    return F.col(colname).isNotNull() & ((F.col(colname) < lo) | (F.col(colname) > hi))

# condición de fila válida (todas las variables en rango si están presentes)
valid_cond = F.lit(True)
for c, (lo, hi) in rules.items():
    valid_cond = valid_cond & (~out_of_range(c, lo, hi))

df_valid = df_ok.filter(valid_cond)
df_invalid_vals = df_ok.filter(~valid_cond)

print("Total:", df_qc.count())
print("Fecha mala:", df_bad_fecha.count())
print("Fecha ok:", df_ok.count())
print("Validos:", df_valid.count())
print("Invalidos por valores:", df_invalid_vals.count())
```

Ver por qué una fila fue inválida

```python
flag_cols = []
for c, (lo, hi) in rules.items():
    flag_cols.append(
        F.when(out_of_range(c, lo, hi), F.lit(f"{c}[{lo},{hi}]")).otherwise(F.lit(None)).alias(f"bad_{c}")
    )

df_invalid_explain = df_invalid_vals.select(
    "estacion_id", "fecha_hora", "fecha_hora_raw",
    "temperatura", "humedad", "ph", "oxigeno_dis", "conductividad", "turbidez",
    "observaciones",
    *flag_cols
)

df_invalid_explain.show(50, truncate=False)
```

Resumen: cuántos outliers por columna

```python
summary_exprs = []
for c, (lo, hi) in rules.items():
    summary_exprs.append(F.sum(out_of_range(c, lo, hi).cast("int")).alias(f"out_{c}"))

df_ok.agg(*summary_exprs).show(truncate=False)
```

filtrado estricto (cualquier variable fuera = se elimina)

```python
from pyspark.sql import functions as F

rules = {
    "temperatura": (0.0, 40.0),
    "humedad": (0.0, 100.0),
    "ph": (0.0, 14.0),
    "oxigeno_dis": (0.0, 20.0),
    "conductividad": (0.0, 10000.0),
    "turbidez": (0.0, 3000.0),
}

def out_of_range(colname, lo, hi):
    return F.col(colname).isNotNull() & ((F.col(colname) < lo) | (F.col(colname) > hi))

# crea strings de motivo; si no aplica, queda NULL
reason_exprs = [
    F.when(out_of_range(c, lo, hi), F.lit(f"{c} fuera [{lo},{hi}]"))
    for c, (lo, hi) in rules.items()
]

df_invalid_report = df_invalid_vals.withColumn(
    "motivo",
    F.concat_ws(" | ", *reason_exprs)   # ignora NULLs
)

df_invalid_report.select(
    "estacion_id", "fecha_hora",
    "temperatura", "humedad", "ph", "oxigeno_dis", "conductividad", "turbidez",
    "motivo", "observaciones"
).show(50, truncate=False)

print("Fecha mala:", df_bad_fecha.count())
print("Fecha ok:", df_ok_fecha.count())
print("Validos:", df_valid.count())
print("Invalidos por valores:", df_invalid_vals.count())
```

Ver rápidamente ejemplos de inválidos y la razón (columna que falló)

```python
reasons = []
for c, (lo, hi) in rules.items():
    reasons.append(
        F.when(out_of_range(c, lo, hi), F.lit(f"{c} fuera [{lo},{hi}]")).otherwise(F.lit(None))
    )

df_invalid_report = df_invalid_vals.withColumn("motivo", F.array_remove(F.array(*reasons), None))

df_invalid_report.select(
    "estacion_id", "fecha_hora",
    "temperatura", "humedad", "ph", "oxigeno_dis", "conductividad", "turbidez",
    "motivo", "observaciones"
).show(50, truncate=False)
```

Reporte QA por estación (cuántos se rechazaron)
ver si una estación está fallando mucho.

```python
from pyspark.sql import functions as F

total_por_est = df_qc.groupBy("estacion_id").count().withColumnRenamed("count","total")

bad_fecha_por_est = df_bad_fecha.groupBy("estacion_id").count().withColumnRenamed("count","bad_fecha")

bad_vals_por_est = df_invalid_vals.groupBy("estacion_id").count().withColumnRenamed("count","bad_vals")

valid_por_est = df_valid.groupBy("estacion_id").count().withColumnRenamed("count","valid")

reporte = (total_por_est
    .join(bad_fecha_por_est, "estacion_id", "left")
    .join(bad_vals_por_est, "estacion_id", "left")
    .join(valid_por_est, "estacion_id", "left")
    .na.fill(0)
    .withColumn("pct_bad_fecha", F.round(F.col("bad_fecha")/F.col("total")*100, 2))
    .withColumn("pct_bad_vals",  F.round(F.col("bad_vals")/F.col("total")*100, 2))
    .withColumn("pct_valid",     F.round(F.col("valid")/F.col("total")*100, 2))
    .orderBy(F.col("pct_bad_vals").desc())
)

reporte.show(truncate=False)
```
